## Initial Data Quality Audit

- Dataset size: 5,110 observations and 12 variables.
- Target variable: `stroke`.
- Missing values detected only in `bmi`: 201 records (3.93%).
- No duplicate rows detected.
- `id` is unique for all records and appears to be an identifier rather than a predictive feature.
- Severe target imbalance:
  - No stroke: 4,861 (95.13%)
  - Stroke: 249 (4.87%)
- `smoking_status` contains an `Unknown` category in 1,544 records (30.22%).
- Numerical variables contain potentially extreme observations, particularly BMI, which require further investigation rather than automatic removal.
- Binary variables `hypertension`, `heart_disease`, and `stroke` contain only valid 0/1 values.
- Age ranges from 0.08 to 82 years, indicating the dataset includes paediatric as well as adult records.

```python
continuous_cols = ["age", "avg_glucose_level", "bmi"]

for col in continuous_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col}")
    print("Lower bound:", round(lower, 2))
    print("Upper bound:", round(upper, 2))
    print("Number of outliers:", len(outliers))
```

## Outlier and Leakage Investigation

```python
df.sort_values("avg_glucose_level", ascending=False)[
    ["age", "gender", "avg_glucose_level", "bmi",
     "hypertension", "heart_disease", "stroke"]
].head(20)
```

```python
df.sort_values("avg_glucose_level")[
    ["age", "gender", "avg_glucose_level", "bmi",
     "hypertension", "heart_disease", "stroke"]
].head(20)
```

```python
q1 = df["avg_glucose_level"].quantile(0.25)
q3 = df["avg_glucose_level"].quantile(0.75)
iqr = q3 - q1

upper_glucose = q3 + 1.5 * iqr

glucose_outliers = df[df["avg_glucose_level"] > upper_glucose]

print("Glucose outlier records:", len(glucose_outliers))
print("Stroke cases among them:", glucose_outliers["stroke"].sum())

print(
    "Stroke rate among glucose outliers:",
    round(glucose_outliers["stroke"].mean() * 100, 2),
    "%"
)
```

```python
q1 = df["bmi"].quantile(0.25)
q3 = df["bmi"].quantile(0.75)
iqr = q3 - q1

upper_bmi = q3 + 1.5 * iqr

bmi_outliers = df[df["bmi"] > upper_bmi]

print("BMI outlier records:", len(bmi_outliers))
print("Stroke cases among them:", bmi_outliers["stroke"].sum())

print(
    "Stroke rate among BMI outliers:",
    round(bmi_outliers["stroke"].mean() * 100, 2),
    "%"
)
```

### Preliminary findings

- IQR analysis detected no age outliers.
- Average glucose level contained 627 observations outside the conventional IQR upper boundary.
- BMI contained 110 observations outside the conventional IQR upper boundary.
- These observations will not be automatically removed, because statistical extremity does not establish data error and clinically extreme values may contain predictive information.
- Stroke-positive participants were substantially older on average and had higher average glucose levels than stroke-negative participants.
- No numerical feature showed an unusually high correlation with the target that would immediately suggest target leakage.
- `id` remains a non-predictive identifier candidate and will be excluded from modelling.
- Categorical stroke-rate differences appear plausible but do not by themselves establish leakage.

### Missingness and validity findings

- No non-positive values were detected for age, average glucose level, or BMI.
- Age ranged from 0.08 to 82 years.
- Average glucose level ranged from 55.12 to 271.74.
- BMI ranged from 10.3 to 97.6.
- BMI was missing in 201 observations (3.93%).
- Stroke prevalence among observations with recorded BMI was approximately 4.26%, whereas stroke prevalence among observations with missing BMI was approximately 19.90%.
- Therefore, complete-case deletion of observations with missing BMI may disproportionately remove positive stroke cases and will not be adopted without further methodological justification.
- The `id` variable showed no obvious relationship with the outcome and represents a unique record identifier; it will not be used as a predictor.

## Final Audit Decisions

- Keep the full dataset for preprocessing.
- Do not remove high glucose or BMI observations solely because they exceed IQR thresholds.
- Treat `smoking_status = "Unknown"` as a legitimate category for now.
- Retain the single `gender = "Other"` observation for now.
- Exclude `id` from predictive modelling.
- Do not use complete-case deletion for missing BMI because missingness is associated with a much higher stroke rate and would remove 40 positive cases.
- Handle BMI missingness inside the preprocessing pipeline to avoid data leakage.
- All transformations, imputation, encoding, scaling, and resampling must be fitted on training data only.
